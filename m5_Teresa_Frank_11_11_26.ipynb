```python
# Module 5 - Houston Weather Analysis
# Teresa Frank

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the Houston Weather dataset

url = "https://raw.githubusercontent.com/papayuh/datn1370-datasets/main/data/houston_weather.csv"

df = pd.read_csv(url)

# Display the first five rows
df.head()
```

```python
# Examine the dataset

print("Dataset size:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
# Prepare the date column

df['date'] = pd.to_datetime(df['date'])

# Create a month column
df['month'] = df['date'].dt.month

# Display the updated dataset
df.head()
```

```python
# Calculate average high temperature by month

monthly_temps = df.groupby('month')['temp_high_f'].mean()

print(monthly_temps)
```

```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.month

df.head()
```

```python
monthly_temps = df.groupby('month')['temp_high_f'].mean()

print(monthly_temps)
```

```python
# Average high temperatures by month

monthly_temps.plot(kind='bar', figsize=(10, 5))

plt.title('Average Daily High Temperature by Month in Houston')
plt.xlabel('Month')
plt.ylabel('Average High Temperature (F)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The bar chart compares Houston's average daily high temperatures across the 12 months. The chart helps identify seasonal temperature patterns and shows which months have higher and lower average temperatures. A bar chart is appropriate because it makes it easy to compare numerical values across separate categories, in this case, the months of the year.

```python
# Distribution of daily high temperatures

plt.figure(figsize=(10, 5))

plt.hist(df['temp_high_f'], bins=15, edgecolor='black')

plt.title('Distribution of Daily High Temperatures in Houston')
plt.xlabel('Daily High Temperature (F)')
plt.ylabel('Number of Days')

plt.tight_layout()
plt.show()
```

The histogram shows the distribution of Houston's daily high temperatures. It helps identify which temperature ranges occur most frequently and how widely temperatures vary throughout the dataset. A histogram is appropriate because it groups numerical data into intervals and makes it easier to see patterns in the distribution of temperatures.
